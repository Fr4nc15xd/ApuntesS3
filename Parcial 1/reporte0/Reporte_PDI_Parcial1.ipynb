## 1.1. Visión Humana: Anatomía y Funcionamiento

La visión humana es un proceso biológico complejo que nos permite interactuar y comprender nuestro entorno. Este proceso comienza con la entrada de luz en el ojo, la cual atraviesa varias estructuras hasta llegar a la retina, donde se transforma en señales eléctricas que el cerebro interpreta como imágenes.

### Anatomía:

1.  **Córnea:** es la membrana transparente frontal del ojo que enfoca inicialmente los rayos luminosos
2.  **Cámara Anterior:** Contiene humor acuoso, por donde los rayos luminosos continúan su trayecto
3.  **Pupila:** su apertura es regulada por el iris, controlando la cantidad de luz que ingresa al ojo
4.  **Cristalino:** Ajustando el enfoque para que la imagen se forme nítidamente en la retina
5.  **Retina:** capa sensible a la luz en la parte posterior del ojo, donde la imagen invertida y real se forma.


La retina alberga dos tipos principales de células fotorreceptoras, encargadas de captar la luz y convertirla en impulsos nerviosos:

*   **Conos:** Son responsables de la visión en color y la percepción de detalles finos. Existen tres tipos de conos, cada uno sensible a diferentes longitudes de onda de luz (rojo, verde y azul), lo que nos permite percibir un amplio espectro de colores. Se concentran principalmente en la fóvea.
*   **Bastones:** Más numerosos que los conos, son cruciales para la visión en condiciones de baja luminosidad (visión nocturna) y para la detección de movimiento. No son sensibles al color.

Cuando la luz estimula estas células fotorreceptoras, se desencadena una serie de reacciones bioquímicas y eléctricas que transforman la energía lumínica en señales neuronales.

### Procesamiento Cerebral:

Estas señales son procesadas por otras células de la retina (bipolares y ganglionares) y, a través del **nervio óptico**, son enviadas a la **corteza visual** del cerebro. Aquí, el cerebro realiza un procesamiento complejo para interpretar y reconstruir las imágenes, lo que incluye:

*   **Reconocimiento de formas y objetos:** Identificación de patrones y estructuras.
*   **Percepción de profundidad:** Determinación de distancias y volúmenes.
*   **Detección de movimiento:** Seguimiento de objetos en desplazamiento.
*   **Constancia del color:** Mantenimiento de la percepción del color de un objeto a pesar de las variaciones de iluminación.

# 2.1. Adquisición y digitalización de imágenes

Es la primera y más fundamental etapa dentro de cualquier sistema de Procesamiento Digital de Imágenes (PDI). Consiste capturar la imagen desde el mundo físico (o desde un archivo) y convertirla en una representación matricial que la computadora pueda manipular. Existen dos formas de obtener esta representación:

Adquisición directa (captura física): un dispositivo como cámara digital, escáner, sensor CCD/CMOS, o incluso equipo médico como resonadores, tomógrafos, traducen la luz proveniente de una escena en señales eléctricas, y estas a su vez en números discretos.


Adquisición indirecta (lectura desde archivo): cuando la imagen ya pasó por el proceso anterior y reside como archivo en un disco para posteriormente convertirlo en un arreglo que el programa pueda manipular.



# 2.2. Rutas


Cuando se carga una imagen desde disco, la forma en que se escribe la ruta del archivo cambia dependiendo del sistema operativo desde el que se trabaje.


**Windows:**

Utiliza la diagonal invertida (\\) como separador entre carpetas,sin embargo, en Python también se usa ese mismo carácter para escribir secuencias de escape. Esto significa que una ruta escrita de forma directa puede romperse si alguna combinación de letras coincide, sin querer, con una secuencia de escape válida:

Para evitar este problema, se antepone una r antes de las comillas, convirtiendo la cadena en un raw string:








```python

# puede generar errores o resultados inesperados: \U y \D podrían interpretarse como escapes inválidos
print("C:\Users\mike\Downloads\siames.jpg")


#forma correcta en Windows: raw string - agregando 'r':
print(r"C:\Users\mike\Downloads\siames.jpg")
```

MacOS y Linux:


Al estar basados en Unix, utiliza la diagonal normal (/) como separador de carpetas y no coincide con el carácter de escape de Python, por tanto, las rutas se pueden escribir de forma directa sin necesidad de usar r"...":




```python
print("/Users/miguelcamargorojas/Downloads/siames.jpg")
python
imagen = plt.imread(r"/Users/miguelcamargorojas/Downloads/siames.jpg")
```

# 2.3. Tipos y formatos de archivos de imágenes digitales


**Formatos bits:** Codifican la imagen como una matriz discreta de píxeles, donde cada elemento almacena un valor de intensidad o color. Su resolución es fija: al escalar la imagen por encima de su tamaño original, se pierde información y aparece pixelado, ya que no existe función de interpolación implícita en el archivo.

Ejemplos:
- JPG
- PNG
- TIFF
- GIF
- BMP




**Formatos vectoriales:**  Codifican la imagen mediante primitivas geométricas (líneas, polígonos) y sus parámetros matemáticos, en lugar de valores de píxel. Son escalables sin pérdida de calidad, ya que se re-renderizan en cada operación de zoom.


Ejemplos:
- EPS
- SVG
- AI
